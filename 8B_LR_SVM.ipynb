# Standardized the data

```python
df_stand = pd.DataFrame(columns=['f1_std', 'f2_std','f3_std']).astype('float64')
```

```python
print(df_stand)
```

```python
f1_std = []
f2_std = []
f3_std = []
for value in data["f1"]:
    value = ((value - data['f1'].mean())/(data['f1'].std()))
    f1_std.append(value)
for value in data["f2"]:
    value = ((value - data['f2'].mean())/(data['f2'].std()))
    f2_std.append(value)
for value in data["f3"]:
    value = ((value - data['f3'].mean())/(data['f3'].std()))
    f3_std.append(value)            
```

```python
df_stand["f1_std"] = f1_std
df_stand["f2_std"] = f2_std
df_stand["f3_std"] = f3_std
```

```python
print(df_stand.head())
```

```python
X_std=df_stand[['f1_std','f2_std','f3_std']].values
print(X_std.shape)
print(Y.shape)
```

# Applying LR on standradized data

```python
clf = SGDClassifier(loss="log", penalty="l2")
clf.fit(X_std, Y)
train = clf.fit(X_std,Y)
yhat = clf.predict(X_std)
print(train.coef_[0],train.intercept_[0])
#draw_line(train.coef_[0],train.intercept_[0],np.min(X_std),np.max(X_std))
#plt.scatter(X_std[:,0],X_std[:,1],c=Y)
```

# Observation

The coefficient on standardized data [-0.19066245  0.3667424  10.5107287 ]  and intercept is -1.876270597276803
