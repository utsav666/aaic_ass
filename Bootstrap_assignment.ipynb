*  <font color='blue'><b> Write code for generating samples</b></font>

```python
import random
from random import randrange
def generating_samples(input_data, target_data):

    '''In this function, we will write code for generating 30 samples '''
    # you can use random.choice to generate random indices without replacement
    # Please have a look at this link https://docs.scipy.org/doc/numpy-1.16.0/reference/generated/numpy.random.choice.html for more details
    # Please follow above pseudo code for generating samples 
    selecting_rows = np.random.choice(input_data.shape[0], int(np.floor(input_data.shape[0] * 0.6)), replace=False) 
    replacing_rows = np.random.choice(selecting_rows.shape[0], int(np.ceil(input_data.shape[0] * 0.4)), replace=False)
    selecting_columns = np.random.choice(input_data.shape[1], np.random.choice(np.arange(3, input_data.shape[1] + 1)), replace=False)
    sample_data = input_data[selecting_rows[:, None], selecting_columns]
    target_sample_data = target_data[selecting_rows]

    replicated_sample_data = sample_data[replacing_rows]
    target_replicated_sample_data = target_sample_data[replacing_rows]

    final_sample_data = np.vstack((sample_data, replicated_sample_data))
    final_target_data = np.vstack((target_sample_data.reshape(-1,1), target_replicated_sample_data.reshape(-1,1)))
    return final_sample_data,final_target_data,selecting_rows,selecting_columns
    # return sampled_input_data , sampled_target_data,selected_rows,selected_columns
    #note please return as lists
a,b,c,d = generating_samples(x,y)    
```

```python
# Use generating_samples function to create 30 samples 
# store these created samples in a list
list_input_data =[]
list_output_data =[]
list_selected_row= []
list_selected_columns=[]

for i in range(0,30):
    a,b,c,d = generating_samples(x,y)
    list_input_data.append(np.array(a))
    list_output_data.append(np.array(b))
    list_selected_row.append(c)
    list_selected_columns.append(d)
```

```python
from sklearn.tree import DecisionTreeRegressor

list_of_all_model = []
for index ,value in enumerate(list_input_data):
    model = DecisionTreeRegressor(max_depth=None)
    model.fit(value,list_output_data[index])
    list_of_all_model.append(model)
    
```

After getting predicted_y for each data point, we can use sklearns mean_squared_error to calculate the MSE between predicted_y and actual_y.

*  <font color='blue'><b> Write code for calculating MSE</b></font>

```python
Y_predicted_value = []
for value in x:
    #print(type(value))
    pred_y_list = []    
    for index, model in enumerate(list_of_all_model):
        pred_y = model.predict(value[list_selected_columns[index]].reshape(1, -1))
        pred_y_list.append(pred_y)
        #print(type(value[list_selected_columns[index]].reshape(1, -1)))
    median_pred_y = np.median(pred_y_list)
    Y_predicted_value.append(median_pred_y)
#print(predicted_y)
mse = mean_squared_error(y, Y_predicted_value)
print(mse)
        
```

Now calculate the $OOB Score =  \frac{1}{506}\sum_{i=1}^{506}(y^{i} - y^{i}_{pred})^{2}$.

*  <font color='blue'><b> Write code for calculating OOB score </b></font>

```python
import warnings
warnings.filterwarnings("ignore")

sq_error=[]
for index, value in enumerate(x):
    all_pred_y = []
    for i, v in enumerate(list_output_data):
        if y[index] not in list_output_data[i]:
            pred_y = list_of_all_model[i].predict(value[list_selected_columns[i]].reshape(1, -1))
            all_pred_y.append(pred_y)               
    median_pred_y = np.median(np.asarray(all_pred_y))
    sq_error.append(np.square(y[index] - median_pred_y))

oob_val = np.nansum(sq_error) / len(x)
print(oob_val)
```

# <font color='blue'><b>Task 2</b></font>

```python
MSE_value = []
OOB_value = []
for index_value in range(0,35):
    list_input_data =[]
    list_output_data =[]
    list_selected_row= []
    list_selected_columns=[]

    for i in range(0,30):
        a,b,c,d = generating_samples(x,y)
        list_input_data.append(np.array(a))
        list_output_data.append(np.array(b))
        list_selected_row.append(c)
        list_selected_columns.append(d)
    list_of_all_model = []
    for index ,value in enumerate(list_input_data):
        model = DecisionTreeRegressor(max_depth=None)
        model.fit(value,list_output_data[index])
        list_of_all_model.append(model)
    Y_predicted_value = []
    for value in x:
        #print(type(value))
        pred_y_list = []    
        for index, model in enumerate(list_of_all_model):
            pred_y = model.predict(value[list_selected_columns[index]].reshape(1, -1))
            pred_y_list.append(pred_y)
            #print(type(value[list_selected_columns[index]].reshape(1, -1)))
        median_pred_y = np.median(pred_y_list)
        Y_predicted_value.append(median_pred_y)
    #print(predicted_y)
    mse = mean_squared_error(y, Y_predicted_value)
    MSE_value.append(mse)    
    sq_error=[]
    for index, value in enumerate(x):
        all_pred_y = []
        for i, v in enumerate(list_output_data):
            if y[index] not in list_output_data[i]:
                pred_y = list_of_all_model[i].predict(value[list_selected_columns[i]].reshape(1, -1))
                all_pred_y.append(pred_y)               
        median_pred_y = np.median(np.asarray(all_pred_y))
        sq_error.append(np.square(y[index] - median_pred_y))

    oob_val = np.nansum(sq_error) / len(x)
    OOB_value.append(oob_val)    
```

```python
mean_MSE = np.asarray(MSE_value).mean()
std_MSE = np.asarray(MSE_value).std()

left_limit_MSE = np.round(mean_MSE - 2*(std_MSE/ np.sqrt(len(MSE_value))), 3)
right_limit_MSE = np.round(mean_MSE + 2*(std_MSE / np.sqrt(len(MSE_value))), 3)

print(mean_MSE,std_MSE,left_limit_MSE,right_limit_MSE)

```

```python
mean_OOB = np.asarray(OOB_value).mean()
std_OOB = np.asarray(OOB_value).std()

left_limit_OOB = np.round(mean_OOB - 2*(std_OOB/ np.sqrt(len(OOB_value))), 3)
right_limit_OOB = np.round(mean_OOB + 2*(std_OOB / np.sqrt(len(OOB_value))), 3)

print(mean_OOB,std_OOB,left_limit_OOB,right_limit_OOB)
```
