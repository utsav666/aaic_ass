```python
df=pd.read_csv('5_a.csv')
df.head(5)
```

```python
def add_column(df):
    df['Y_pred'] = df['proba'].apply(lambda x: 1 if x > 0.5 else 0)
    return df

```

# Task 1 for 5_a.csv

# Calculate Confusion Matrix

```python
def confusion_matrix(df):
    True_Positive= ((df['y']==1) & (df['Y_pred']==1)).sum()
            
    True_Negative = ((df['y']==0) & (df['Y_pred']==0)).sum()
            
    False_Negative = ((df['y']==1) & (df['Y_pred']==0)).sum()
            
    False_Positive = ((df['y']==0) & (df['Y_pred']==1)).sum()
            
    return True_Positive ,True_Negative ,False_Negative,False_Positive




```

# Calculate  F1 SCORE

```python
def compute_f1_score(TP, FP, FN):
    
    precision = TP / (TP + FP)
    recall = TP / (TP + FN)
    f1_score = (2*precision*recall)/(precision + recall)
    return precision, recall, f1_score
```

# Calculate Accuracy Score

```python
def accuracy_score(TP, TN, FP, FN):
    accu_score = (TP + TN) / (TP + TN + FP + FN)
    print(accu_score,"score")
    return accu_score
```

# Calculate Auc Score

```python
def auc_score(df):
    tpr = []
    fpr = []
    for threshold in (df["proba"].unique()):
        df['y_hat']  = np.where(df["proba"]>=threshold,1,0)
        True_Positive= ((df['y']==1) & (df['y_hat']==1)).sum()
            
        True_Negative = ((df['y']==0) & (df['y_hat']==0)).sum()
            
        False_Negative = ((df['y']==1) & (df['y_hat']==0)).sum()
            
        False_Positive = ((df['y']==0) & (df['y_hat']==1)).sum()
        tpr.append(True_Positive/(True_Positive+False_Negative))
        fpr.append(False_Positive/(False_Positive+True_Negative))
    return tpr,fpr    


```

```python
df = add_column(df)
```

# Printing confusion matrix

```python
TP,TN,FN,FP=confusion_matrix(df)
print("CONFUSION MATRIX**********************************************************************************")
print(TP,"True Positive")
print(TN,"True Negative")
print(FN,"False Negative")
print(FP,"False Positive")
```

# Printing precision, recall, f1_score

```python
precision, recall, f1_score=compute_f1_score(TP, FP, FN)
print("Precision:", precision,"Recall:", recall, "F1 Score :", f1_score)
```

# Printing AUC Score 

```python
df.sort_values("proba", inplace=True,ascending = False)
tpr_array , fpr_array = auc_score(df)
auc=np.trapz(np.asarray(tpr_array) , np.asarray(fpr_array))
print(auc,"auc score ")
```

# Printing accuracy score 

```python
accuracy_score_1 = accuracy_score(TP, TN, FP, FN)
print(accuracy_score_1)
```

# Calculating MSE

```python
df_D=pd.read_csv('5_d.csv')
count_of_y=list(df_D['y'])
count_of_pred=list(df_D['pred'])
sumtion = 0  
n = len(count_of_y) 
for i in range (0,n):  
    diff = count_of_y[i] - count_of_pred[i]  
    sq_diff = diff**2  
    sumtion = sumtion + sq_diff  
    MSE = sumtion/n  
print("The Mean Square Error is: " , MSE)
```

# Calculating MAPE

```python
y_matrix=df_D['y']
pred_matrix=df_D['pred']
def mean_abs_percentage_error(y_matrix, pred_matrix): 
    y, pred_y = np.array(y_matrix), np.array(pred_matrix)
    return np.mean(np.abs((y - pred_y) / np.mean(y)))

mean_abs_percentage_error(y_matrix, pred_matrix)
```

# Calculating R Square Error

```python
def RSqr(y_matrix, pred_matrix):
    sse_val = sum((y_matrix - pred_matrix)**2)
    tse_val = (len(y_matrix) - 1) * np.var(y_matrix, ddof=1)
    score_of_r2 = 1 - (sse_val / tse_val)
    return score_of_r2

r_sqr_error = RSqr(y_matrix, pred_matrix)
print(r_sqr_error)
```
